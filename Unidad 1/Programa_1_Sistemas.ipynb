# **Listas en phyton**

```python
alumnos = ["Ana", "Luis", " Carlos"]
print(alumnos)
print(alumnos[0])
print(alumnos[2])
```

```python
# El metodo apped agrega un elemento al final de la lista
print(alumnos)
alumnos.append("Sandra")
print(alumnos)
```

```python
print(alumnos)
alumnos.insert(2,"Sofia")
print(alumnos)
```

```python
# Remove elimina un elemento de la lista especificando el
# valor, si existen valores repetidos, borra el primer valor
# que encuentra
print(alumnos)
alumnos.remove("Sofia")
print(alumnos)
```

```python
# La siguiente linea marca error porque queremos acceder a
# un indice que no existe, la lista tiene unicamente 6 elementos
print(alumnos[2])
print(alumnos[7])
```

```python
# pop: me permite eliminar un elemento de la lista especificando su posicion
print(alumnos)
alumnos.pop(2)
print(alumnos)
```

```python
# del: me permite eliminar un elemento de la lista especificando su posicion = pop
print(alumnos)
del alumnos[2]
print(alumnos)
```

```python
# len: me permite conocer el numero de elementos de la lista
print("numero de elementos o items de la lista", len(alumnos))
```

```python
# Ciclo para recorrer la lista y hacer algun proceso
alumnos.append("Sofia")
alumnos.append("Carlos")
for alumno in alumnos:
  print(alumno)
  if alumno == "Sofia":
    print("Se encontro a Sofia")
    break # forza a salir del ciclo
```
