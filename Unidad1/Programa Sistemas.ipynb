```python
# Ejemplo de programa con multiples condiciones, cuando la primera no se cumple,
# Se vuelve a realizar otra condicion y si no se cumple, se revisa otra condicion y asi
# sucesivamente las veces que sean necesarias de acuerdo a las necesidades
edad = int(input("Dame tu edad"))
if edad <= 10:
  print("Eres un niño")
elif edad <= 18:
  print("Eres un adolescente")
elif edad <= 30:
  print("Eres un adulto joven")
elif edad <= 65:
  print("Eres un adulto")
else:
  print("Eres un adulto mayor")
```
