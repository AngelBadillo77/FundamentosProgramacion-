```python
from datetime import datetime

# ========================================
# SISTEMA DE PRÉSTAMOS
# Versión 1 - Programación básica y uso de listas
# ========================================

# Lista de alumnos modificada: [número de control, nombre, carrera]
alumnos = [
    {"num_control":"26040200", "nombre": "Superman",    "carerra":  "Ingeniería en Inteligencia Artificial"      },
    {"num_control":"26040201", "nombre": "Bat Man",     "carerra":  "Ingeniería en Inteligencia Artificial"      },
    {"num_control":"26040202", "nombre": "Pedro Pérez", "carerra":  "Ingeniería en Sistemas Computacionales"     },
    {"num_control":"26040203", "nombre": "Ana Gómez",   "carerra":  "Ingeniería en Tecnologías de la Información"},
    {"num_control":"26040204", "nombre": "Luis Torres", "carerra":  "Ingeniería en Sistemas Computacionales"     }
]

# Cada valor en la lista de computadoras tiene su lista correspondiente con el
# aula donde está ubicada, el software instalado y el estado de esa computadora
computadoras = [
    {"id":"PC-LC1-01","aula":"LC1","programas":"Python, VS Code",      "estado":"Disponible"    },
    {"id":"PC-LC1-02","aula":"LC1","programas":"Python, Java, VS Code","estado":"Prestada"      },
    {"id":"PC-LC1-03","aula":"LC1","programas":"AutoCAD, MATLAB",      "estado":"Reservada"     },
    {"id":"PC-LC2-01","aula":"LC2","programas":"Python, VS Code",      "estado":"Disponible"    },
    {"id":"PC-LC2-02","aula":"LC2","programas":"Python, Java, VS Code","estado":"Prestada"      },
    {"id":"PC-LC2-03","aula":"LC2","programas":"AutoCAD, MATLAB",      "estado":"Disponible"    },
    {"id":"PC-LC3-01","aula":"LC3","programas":"Python, VS Code",      "estado":"Disponible"    },
    {"id":"PC-LC3-02","aula":"LC3","programas":"Python, Java, VS Code","estado":"Prestada"      },
    {"id":"PC-LC3-03","aula":"LC3","programas":"AutoCAD, MATLAB",      "estado":"Reservada"     },
    {"id":"PC-LC3-04","aula":"LC3","programas":"Python, VS Code",      "estado":"Mantenimiento" }
]

# Lista para almacenar el historial de préstamos realizados
historial_prestamos = [
    {"num_control":"26040200", "programas":"Python",  "aula":"LC1", "id":"PC-LC1-01",    "fecha":"12/10/2026 12:10:32" },
    {"num_control":"26040201", "programas":"Java",    "aula":"LC1", "id":"PC-LC1-01",    "fecha":"12/10/2026 12:10:32" },
    {"num_control":"26040202", "programas":"MATLAB",  "aula":"LC1", "id":"PC-LC1-01",    "fecha":"12/10/2026 12:10:32" },
    {"num_control":"26040203", "programas":"VS Code", "aula":"LC1", "id":"PC-LC1-01",    "fecha":"12/10/2026 12:10:32" },
    {"num_control":"26040204", "programas":"Python",  "aula":"LC1", "id":"PC-LC1-01",    "fecha":"12/10/2026 12:10:32" }
]

# Lista para almacenar el historial de necesidades de software
historial_necesidades = []


def listar_computadoras():
    print("\n--- COMPUTADORAS ---")
    for i in range(len(computadoras)):
        print(i+1, "Aula:", computadoras[i]["aula"], "| ID:", computadoras[i]["id"],
              "| Software:", computadoras[i]["programas"], " | Estado:", computadoras[i]["estado"])


def buscar_sw():
    programa = input("\nIngrese el software que desea buscar: ")
    print("\n--- RESULTADOS ---")
    encontrado = False
    for i in range(len(computadoras)):
        if programa.lower() in computadoras[i]["programas"].lower():
            print(i+1, "Aula:", computadoras[i]["aula"], "ID:", computadoras[i]["id"],
                  "Software:", computadoras[i]["programas"], "Estado:", computadoras[i]["estado"])
            encontrado = True
    if not encontrado:
        print("No se encontraron computadoras con ese software.")


def solicitar_prestamo():
    num_control = input("Número de control del alumno: ")

    alumno_encontrado = False
    for i in range(len(alumnos)):
        if alumnos[i][0] == num_control:
            alumno_encontrado = True
            break

    if alumno_encontrado:
        programa = input("¿Qué programa necesita?: ")
        computadora_asignada = False

        for i in range(len(computadoras)):
            if programa.lower() in computadoras[i][2].lower():
                if computadoras[i][3] == "Disponible":
                    computadoras[i][3] = "Prestada"

                    # Registrar en el historial: [Alumno, Software, Aula, ID]
                    historial_prestamos.append([num_control, programa, computadoras[i][1], computadoras[i][0]])

                    print("Se asignó la computadora", computadoras[i][0],
                          "que está en el aula", computadoras[i][1])
                    computadora_asignada = True
                    break

        if not computadora_asignada:
            existe_software = False
            for i in range(len(computadoras)):
                if programa.lower() in computadoras[i][2].lower():
                    existe_software = True
                    break

            if existe_software:
                print("Lo sentimos, las computadoras con ese software se encuentran ocupadas (Prestadas, Reservadas o en Mantenimiento).")
            else:
                print("No se encontró ningún software con ese nombre en el sistema.")

                respuesta = input("¿Desea agregarlo a la lista de necesidades? (s/n): ")
                if respuesta.lower() in ["s", "si"]:
                    fecha_hora_actual = datetime.now().strftime("%d-%m-%Y %H:%M:%S")
                    historial_necesidades.append([programa, fecha_hora_actual])
                    print("Programa agregado a la lista de necesidades correctamente.")

    else:
        print("El alumno no existe en la lista.")


def devolver_computadora():
    print("\n=== DEVOLVER COMPUTADORA ===")
    id_pc = input("Dame el ID de la computadora: ")
    existe = False
    for i in range(len(computadoras)):
        if computadoras[i][0] == id_pc:
            existe = True
            if computadoras[i][3] in ["Prestada", "Reservada"]:
                computadoras[i][3] = "Disponible"
                print("La computadora", id_pc, "ya está disponible.")
                break
            else:
                print("La computadora", id_pc, "no está prestada/reservada.")
    if not existe:
        print("No existe la computadora", id_pc)


def agregar_computadora():
    print("\n=== AGREGAR COMPUTADORA ===")
    id_computadora = input("ID de la computadora: ")
    aula = input("Aula: ")
    programas = input("Software instalado: ")
    nueva_compu = [id_computadora, aula, programas, "Disponible"]
    computadoras.append(nueva_compu)
    print("\nComputadora agregada correctamente.")


def mostrar_historial():
    print("\n--- Historial de préstamos de computadoras ---")
    if len(historial_prestamos) == 0:
        print("No hay préstamos registrados en el historial.")
    else:
        for i in range(len(historial_prestamos)):
            print(i+1, "| Alumno:", historial_prestamos[i][0],
                  "| Software:", historial_prestamos[i][1],
                  "| Aula:", historial_prestamos[i][2],
                  "| ID:", historial_prestamos[i][3])


def mostrar_necesidades():
    print("\n--- HISTORIAL DE NECESIDADES ---")
    if len(historial_necesidades) == 0:
        print("No hay necesidades registradas.")
    else:
        for i in range(len(historial_necesidades)):
            print(i+1, "| Programa:", historial_necesidades[i][0], "| Fecha:", historial_necesidades[i][1])


def mostrar_estadisticas():
    if len(historial_prestamos) == 0:
        print("\nNo hay préstamos registrados para generar estadísticas.")
        return

    # --- Conteo por programa ---
    conteo_programas = []
    for i in range(len(historial_prestamos)):
        programa = historial_prestamos[i][1].lower()
        encontrado = False

        for k in range(len(conteo_programas)):
            if conteo_programas[k][0] == programa:
                conteo_programas[k][1] += 1
                encontrado = True
                break

        if not encontrado:
            conteo_programas.append([programa, 1])

    # Mostrar resultados por programa
    print("\n\n====================================")
    print("\nEstadísticas por programa:")
    for i in range(len(conteo_programas)):
        print(conteo_programas[i][0], ":", conteo_programas[i][1], "veces")

    # --- Conteo por carrera ---
    conteo_carreras = []
    for i in range(len(historial_prestamos)):
        num_control = historial_prestamos[i][0]
        carrera = "Desconocida"

        # Buscar la carrera correspondiente al número de control del alumno
        for j in range(len(alumnos)):
            if alumnos[j][0] == num_control:
                carrera = alumnos[j][2]
                break

        encontrado = False
        for k in range(len(conteo_carreras)):
            if conteo_carreras[k][0] == carrera:
                conteo_carreras[k][1] += 1
                encontrado = True
                break

        if not encontrado:
            conteo_carreras.append([carrera, 1])

    # Mostrar resultados por carrera
    print("\n\n====================================")
    print("\nEstadísticas por carrera:")
    for i in range(len(conteo_carreras)):
        print(conteo_carreras[i][0], ":", conteo_carreras[i][1], "veces")


# MENÚ PRINCIPAL
opcion = 0
while opcion != 10:
    print("\n=============================")
    print("      SISTEMA DE PRÉSTAMOS")
    print("=============================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar préstamo")
    print("4. Devolver computadora")
    print("5. Crear reservación")
    print("6. Agregar computadora")
    print("7. Historial de préstamos")
    print("8. Necesidades de software")
    print("9. Estadísticas")
    print("10. Salir")
    print("=============================")

    try:
        opcion = int(input("Seleccione una opción: "))
    except ValueError:
        print("\nPor favor, ingrese un número válido.")
        continue

    if opcion == 1:
        listar_computadoras()
    elif opcion == 2:
        buscar_sw()
    elif opcion == 3:
        solicitar_prestamo()
    elif opcion == 4:
        devolver_computadora()
    elif opcion == 5:
        print("\n[Opción 5 - Reservaciones pendiente de implementar]")
    elif opcion == 6:
        agregar_computadora()
    elif opcion == 7:
        mostrar_historial()
    elif opcion == 8:
        mostrar_necesidades()
    elif opcion == 9:
        mostrar_estadisticas()
    elif opcion == 10:
        print("\nGracias por utilizar el sistema.")
    else:
        print("\nOpción no válida.")
```
